# 06 — HMM / Viterbi Route Reconstruction

Use classifier probabilities as emission scores and directed road-network connectivity as a transition preference. The decoder restarts after long timestamp gaps.

```python
from pathlib import Path
import sys, os
ROOT = Path.cwd()
if not (ROOT / "src").exists(): ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT / "src"))
DATA = ROOT / "data"
import joblib
from pathsnap import read_trip, make_candidates, viterbi_decode
trip = sorted((DATA / "student_test").iterdir())[0]
track, nodes, arcs, route = read_trip(trip)
X,y,obs,cids,truth,miss,*_ = make_candidates(track,nodes,arcs,route)
model = joblib.load(ROOT / "models" / "model.joblib")
predicted_arc_ids = viterbi_decode(model,X,obs,cids,arcs,timestamps=track[:,2])
print("Trip:",trip.name,"GPS points:",len(track),"decoded arcs:",len(predicted_arc_ids),"candidate misses:",miss)
```
