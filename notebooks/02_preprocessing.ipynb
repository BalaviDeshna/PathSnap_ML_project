# 02 — Preprocessing

Load raw trip files and inspect coordinate projection and timestamp-derived quantities. Large timestamp gaps are treated as recording interruptions by the decoder.

```python
from pathlib import Path
import sys, os
ROOT = Path.cwd()
if not (ROOT / "src").exists(): ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT / "src"))
DATA = ROOT / "data"
from pathsnap import read_trip, local_xy
trip = sorted((DATA / "train").iterdir())[0]
track, nodes, arcs, route = read_trip(trip)
print("Trip:", trip.name)
print("track/nodes/arcs/route:", track.shape, nodes.shape, arcs.shape, route.shape)
xy = local_xy(track[:,0], track[:,1], track[:,0].mean(), track[:,1].mean())
print("Projected GPS coordinates:", xy.shape)
```
